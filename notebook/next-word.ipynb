```python
document = """
THE KEEPER OF SMALL THINGS
A Story of History and Culture Around the World


PROLOGUE: THE CEDAR CHEST

The rain had been falling on the harbor town of Port Selene for three days, and Noor was bored. She had read every comic book in the house twice, counted the cracks in the kitchen ceiling (there were seventeen), and argued with the cat about whose chair was whose. The cat had won.

"Grandfather," she said at last, flopping onto the rug in his study, "there's nothing to do. Nothing in the whole world."

Elias Marlow looked up from his desk. He was seventy-eight years old, with a white beard trimmed short and spectacles that slid down his nose whenever he laughed. For forty years he had worked as a museum curator, then as a teacher, and finally as what he called "a professional wanderer." His study smelled of old paper, pipe tobacco he no longer smoked, and the faint sweetness of cedar.

"Nothing in the whole world?" he repeated. "That's a very large claim, Noor. The world is a crowded place. It has roughly eight billion people in it, about seven thousand living languages, and more stories than anyone could hear in a thousand lifetimes."

"But they're all out there," she said, waving at the rain-streaked window. "And I'm in here."

Elias smiled. He rose slowly, crossed the room, and knelt beside a long wooden chest that had always sat beneath the bookshelves. Noor had asked about it before, but he'd always said, "Not yet." Now he lifted the brass latch and opened the lid.

Inside were dozens of small objects, each one resting in its own compartment lined with faded blue velvet: a clay tablet no bigger than a biscuit, a carved bead, a scrap of silk, a coin worn almost smooth, a brass instrument covered in tiny numbers, a knotted bundle of colored strings, a seashell, a folded railway ticket. Every compartment had a paper label written in his careful, slanting hand.

"These," Elias said, "are my small things. I collected some of them on my travels. Others were given to me by friends, students, and strangers. A few are replicas, because the real ones belong in museums or, better still, in the places they came from. But every single one of them is a door."

"A door to where?"

"To everywhere," he said. "And to everywhen. Pick one."

Noor reached in and lifted out the little clay tablet. It was covered in rows of tiny wedge-shaped marks, like the footprints of a bird that had walked carefully in straight lines.

"Ah," said Elias, settling back into his armchair. "You've chosen the beginning. Or one of them, at least. Sit down, and I'll tell you about the land between the rivers."


CHAPTER ONE: THE LAND BETWEEN THE RIVERS

"Long ago," Elias began, "about five and a half thousand years before you were born, people living in a place called Sumer had a problem. They were very good farmers. Too good, in a way."

"How can you be too good at farming?"

"When you grow more barley than you can eat, you have to store it. When you store it, someone has to keep track of how much there is, who owns it, and who owes what to whom. Sumer lay in southern Mesopotamia, a word that comes from Greek and means 'between the rivers.' The rivers were the Tigris and the Euphrates, in what is now Iraq. Their floods brought rich silt to the fields, and the Sumerians dug canals to carry the water even farther. Cities grew up along those canals: Ur, Uruk, Lagash, Eridu. Uruk may have been home to forty thousand people or more by around 3000 BCE, which made it one of the largest cities on Earth at the time."

He tapped the tablet in Noor's hands.

"At first, the record-keepers used little clay tokens: a cone for a measure of grain, a sphere for another, a cylinder for an animal. Later they sealed the tokens inside clay envelopes, and later still they realized they didn't need the tokens at all. They could simply draw signs into wet clay with a cut reed. By about 3200 BCE, this had become writing. Because the reed left wedge-shaped marks, we call the script cuneiform, from the Latin word cuneus, meaning 'wedge.'"

"So writing was invented for... accounting?" Noor wrinkled her nose.

Elias laughed. "Disappointing, isn't it? But give it a few centuries. Once people could write down numbers, they began to write down everything else: laws, letters, prayers, recipes, medical advice, complaints. One of my favorite tablets is from around 1750 BCE. A man named Nanni wrote to a merchant called Ea-nasir, furious that he'd been sold poor-quality copper and treated rudely. Scholars sometimes call it the oldest customer complaint in the world."

"That's actually funny."

"History is often funny, if you listen closely. And sometimes it's beautiful. The Sumerians and the peoples who came after them, the Akkadians, Babylonians, and Assyrians, wrote one of the oldest great works of literature we know: the Epic of Gilgamesh. It tells of a proud king of Uruk who loses his dearest friend, Enkidu, and sets out to find the secret of eternal life. He fails, of course. But on his way home he learns that what lasts is not a single person's life but the walls we build, the cities we share, and the stories we leave behind. People were telling that story more than four thousand years ago, and we're still reading it."

Noor turned the tablet over. "Did they have rules? Like laws?"

"They did. The most famous collection comes from Hammurabi, a king of Babylon who ruled in the eighteenth century BCE. Around 1754 BCE, his laws were carved onto a tall stone pillar of black basalt. There are 282 of them. Some deal with trade, some with farming, some with marriage and inheritance. Many follow the principle of 'an eye for an eye,' though the punishments depended a great deal on whether you were rich, poor, or enslaved. The pillar was discovered in 1901 and now stands in the Louvre Museum in Paris."

He leaned forward, his voice dropping as though sharing a secret.

"The Mesopotamians gave us something else you use every single day. They counted in base sixty. That's why there are sixty seconds in a minute and sixty minutes in an hour, and why a circle has 360 degrees."

"Wait. Every time I look at a clock, I'm looking at something from Babylon?"

"Every single time," said Elias, delighted.

"What about Egypt?" Noor asked. "We learned about pyramids at school."

"Ah, Egypt. To the west of Mesopotamia, across deserts and the narrow land bridge of Sinai, another great civilization grew along another great river: the Nile. The Greek historian Herodotus called Egypt 'the gift of the Nile,' and he was right. Every summer, the river flooded and left behind a layer of black, fertile mud. The Egyptians called their land Kemet, 'the Black Land,' after that soil, and they called the desert Deshret, 'the Red Land.'

"Around 3100 BCE, Upper and Lower Egypt were united under a single ruler. Tradition names him Narmer, or Menes. For nearly three thousand years afterward, Egypt was governed by a long succession of pharaohs, divided by historians into about thirty dynasties. The Great Pyramid of Giza was built for the pharaoh Khufu around 2560 BCE. It originally stood about 146 meters tall, and for more than three thousand eight hundred years it was the tallest human-made structure on the planet."

"Were they built by slaves?"

"That's what many people believed for a long time. But archaeologists have found the villages where the workers lived, with bakeries, breweries, and cemeteries. The evidence suggests most of the builders were paid laborers, skilled craftsmen and farmers who worked in rotating teams, perhaps during the flood season when the fields were underwater. They were fed bread, beer, and quite a lot of meat. Some of them even left graffiti naming their work gangs, with names like 'Friends of Khufu.'"

Noor grinned. "Like a sports team."

"Very much like a sports team. The Egyptians had their own writing too: hieroglyphs, a script of pictures that could stand for sounds, words, or ideas. For centuries after the last hieroglyphic inscription was carved, in 394 CE, nobody could read them. Then, in 1799, French soldiers in the town of Rashid, which Europeans called Rosetta, dug up a broken slab of stone. On it, the same decree from 196 BCE was written three times: in hieroglyphs, in a later Egyptian script called Demotic, and in ancient Greek. Since scholars could read Greek, the stone became a key. In 1822, a young Frenchman named Jean-François Champollion finally cracked the code. Legend says he ran into his brother's office, shouted 'Je tiens l'affaire!' — 'I've got it!' — and fainted."

"Did he really faint?"

"So the story goes. Stories have a way of improving with age, like cheese. But the point is true: a single stone opened a door that had been locked for fourteen centuries."

Noor placed the tablet carefully back in its velvet compartment. "Who's next?"


CHAPTER TWO: CITIES OF BRICK AND THE WHEEL OF LAW

Her fingers found a small reddish bead, long and thin, with a faint pattern of white lines etched into its surface.

"Carnelian," said Elias. "From the Indus Valley. Or a careful copy of one. Beads like this traveled all the way from South Asia to Mesopotamia more than four thousand years ago. They've been found in the royal tombs at Ur."

"So they were trading? Back then?"

"Constantly. The Indus Valley Civilization, sometimes called the Harappan Civilization, flourished in what is now Pakistan and northwestern India. Its mature phase lasted from about 2600 to 1900 BCE. At its height it covered a larger area than either Egypt or Mesopotamia. Its two most famous cities are Harappa and Mohenjo-daro, and they're remarkable for one reason above all others: planning."

"Planning?"

"Streets laid out in a grid. Houses built from baked bricks of standard sizes, in a ratio of one to two to four, whether you were in one city or another hundreds of kilometers away. Covered drains running along the streets. Many houses had their own bathing areas and even toilets connected to the drains. In Mohenjo-daro there's a structure archaeologists call the Great Bath, a watertight pool about twelve meters long, sealed with a layer of natural tar. Some scholars think it was used for ritual washing."

"That sounds more modern than some places now."

"In some ways it was. And here's the mystery: we can't read their writing. The Indus people left thousands of small stone seals with short inscriptions and lovely carvings of animals, especially a creature that looks like a one-horned bull. But the inscriptions are very short, there's no bilingual stone like Rosetta, and we don't know what language they spoke. The Indus script remains undeciphered to this day."

"Maybe I'll decipher it," said Noor.

"I sincerely hope you do. Many have tried."

Elias took off his spectacles and polished them on his sleeve.

"After the Indus cities declined, possibly because of climate change and shifting rivers, the center of life in South Asia gradually moved east toward the plains of the Ganges. This was the age of the Vedas, ancient hymns in Sanskrit, passed down by memory for many centuries before they were written. Reciters trained for years, learning the hymns forward, backward, and in complex interwoven patterns so that not a single syllable would be lost. UNESCO now recognizes that tradition of Vedic chanting as part of humanity's intangible cultural heritage.

"Around the fifth century BCE, a prince named Siddhartha Gautama left his palace in search of an answer to human suffering. After years of searching, he sat beneath a fig tree at a place now called Bodh Gaya and, as Buddhists believe, attained enlightenment. He became known as the Buddha, the Awakened One. His teachings would later spread across Asia, to Sri Lanka, Tibet, China, Korea, Japan, and Southeast Asia."

"And did kings listen to him?"

"One king in particular. His name was Ashoka, and he ruled the Mauryan Empire in the third century BCE. Early in his reign, around 261 BCE, he waged a brutal war against the region of Kalinga, on India's eastern coast. Tens of thousands died. Afterward, Ashoka was overwhelmed by remorse. He turned to Buddhism and began to rule according to dhamma, a word meaning something like righteousness, duty, or moral law. He had edicts carved on rocks and polished stone pillars across his empire, urging people to be tolerant of other religions, kind to animals, and respectful to parents and elders. He built wells and rest houses along the roads, and planted shade trees for travelers."

"That's a very good apology."

"It's one of the most famous changes of heart in history. The lion capital that once crowned one of his pillars, at Sarnath, is now the national emblem of India, and the wheel from that pillar appears at the center of the Indian flag.

"South Asia also gave the world some extraordinary ideas. Mathematicians there developed the decimal place-value system, the one where the position of a digit tells you its value. In 628 CE, the astronomer Brahmagupta wrote down rules for calculating with zero as a number in its own right. These numerals later traveled west through the Islamic world, which is why we call them Hindu-Arabic numerals. The game of chess also seems to have begun in India, around the sixth century CE, as a game called chaturanga, which means 'four divisions,' after the four parts of an ancient army: infantry, cavalry, elephants, and chariots.

"And there was a great center of learning called Nalanda, in what is now the state of Bihar, founded in the fifth century CE. At its height, thousands of students and teachers lived and studied there, some traveling from as far away as China and Korea. The Chinese monk Xuanzang studied at Nalanda in the seventh century and wrote admiring descriptions of its towers and libraries. It flourished for roughly seven hundred years before it was sacked and burned around 1193."

Noor was quiet for a moment. "What about now? What do people celebrate there now?"

"Oh, so much. Diwali, the festival of lights, when homes are filled with clay lamps called diyas to celebrate the victory of light over darkness. Holi, the spring festival of colors, when people throw bright powders at one another, and for one day even strangers become friends. Eid, Christmas, Vaisakhi, Durga Puja, Onam... South Asia is home to nearly a quarter of humanity and a dizzying number of languages, religions, foods, and festivals. You could spend your whole life there and still be surprised every week."


CHAPTER THREE: BONES, BRONZE, AND PAPER

The next object was a small cicada carved from pale green stone, smooth and cool as water.

"Jade," Elias said. "In ancient China, jade was treasured above gold. It was thought to represent purity, wisdom, and endurance. A cicada like this one, which spends years underground before emerging into the light, was a symbol of rebirth. Small jade cicadas were sometimes placed in the mouths of the dead."

Noor nearly dropped it. "Is this from a—"

"No, no," he laughed. "This one was carved in a workshop for a gift shop. I promise."

She relaxed, and he began.

"China has one of the longest continuous written traditions in the world. The earliest Chinese writing that we know of appears on what we call oracle bones, from the Shang dynasty, around 1250 BCE. Kings wanted to know the future: Will the harvest be good? Will it rain? Should we go to war? Will the queen's childbirth go well? Diviners carved these questions onto turtle shells and the shoulder blades of oxen, then applied heat until the bone cracked. They read the answer in the cracks, and often they carved the answer, and sometimes what actually happened, onto the bone as well. Many of the characters on those bones are the distant ancestors of characters still used by more than a billion people today."

"How did anyone find them?"

"There's a lovely story, possibly too lovely to be entirely true, that in 1899 a scholar named Wang Yirong was ill and bought a medicine called 'dragon bones' from a pharmacy. Before they were ground into powder, he noticed ancient writing scratched onto them. Whatever the exact truth, by the early twentieth century scholars had traced the bones to the old Shang capital near Anyang, and tens of thousands have since been excavated.

"After the Shang came the Zhou dynasty, and during its later, troubled centuries, a time when rival states fought constantly, some of China's greatest thinkers appeared. The most influential was Kong Fuzi, whom we call Confucius. He lived from 551 to 479 BCE. He taught that society works best when people cultivate virtue, honor their families, respect learning, and fulfill their responsibilities to one another: rulers to subjects, parents to children, friends to friends. His sayings were collected by his students in a book called the Analects. Around the same era, according to tradition, a sage named Laozi gave rise to Daoism, which teaches harmony with the Dao, the natural 'way' of the universe.

"In 221 BCE, the ruler of the state of Qin conquered all his rivals and declared himself Qin Shi Huang, the First Emperor. He standardized weights, measures, coins, the width of cart axles, and the written script. He connected earlier defensive walls into a long northern barrier. Most of the Great Wall you see in photographs today, though, was built much later, during the Ming dynasty, between the fourteenth and seventeenth centuries."

"And he was buried with an army," Noor said. "I've seen pictures."

"The Terracotta Army. In 1974, farmers digging a well near the city of Xi'an struck pieces of clay. They had found one of the greatest archaeological discoveries of the twentieth century: thousands of life-sized clay soldiers, along with horses, chariots, and weapons, arranged in battle formation to guard the emperor in the afterlife. Archaeologists estimate there may be around eight thousand soldiers, and no two faces are exactly alike. The emperor's own burial mound still hasn't been opened. Ancient accounts claim it contains rivers of mercury, and soil tests around the mound have indeed shown unusually high levels of mercury."

"Creepy," said Noor, with great satisfaction.

"The Han dynasty, which followed soon after, lasted for about four hundred years, from 206 BCE to 220 CE. It was so important that the majority ethnic group in China still calls itself the Han people. Under the Han, a court official named Cai Lun reported a method of making paper to the emperor in 105 CE, using tree bark, hemp, old rags, and fishing nets. Simpler kinds of paper had existed before, but his method made it cheaper and better. Think about that for a moment. Nearly every book, letter, newspaper, and school notebook in history owes something to that innovation.

"Later dynasties added more. During the Tang dynasty, from 618 to 907, the capital Chang'an was perhaps the largest city in the world, with merchants, monks, and diplomats arriving from Persia, Japan, Korea, India, and Central Asia. It was a golden age of poetry, the age of Li Bai and Du Fu, whose poems Chinese schoolchildren still memorize. The oldest dated printed book that survives is a copy of the Diamond Sutra, a Buddhist text printed with carved wooden blocks in 868 CE. It was found in a sealed cave near Dunhuang in 1900. And during the Song dynasty, around the 1040s, a craftsman named Bi Sheng invented movable type made of baked clay, some four centuries before Gutenberg.

"The Chinese also gave the world the compass, gunpowder, porcelain, silk, and tea. And one of the world's most widely celebrated holidays: the Lunar New Year, or Spring Festival. Families travel enormous distances to be together, in what's often called the largest annual human migration on Earth. There are red lanterns, red envelopes with money for children, dumplings shaped like ancient gold ingots, and firecrackers to frighten away a legendary monster called Nian."

"A monster?"

"A monster who, according to legend, was afraid of loud noises and the color red. Which is very convenient, if you think about it, because it means the best way to defeat him is a party."


CHAPTER FOUR: THE ROADS OF SILK AND SPICE

Noor pulled out a scrap of golden silk, embroidered with tiny birds. It was frayed at the edges and so light it seemed to float in her palm.

"For thousands of years," Elias said, "China guarded the secret of making silk. The cloth came from the cocoons of silkworms that fed on mulberry leaves, and revealing how it was made could be punishable by death. People in Rome were so eager for it that some Roman writers complained the empire was draining its gold to pay for transparent fabrics."

"So they traded it along the Silk Road."

"Along the Silk Roads. Plural. That's an important detail. It wasn't one road, and nobody back then called it the Silk Road. That name was popularized in 1877 by a German geographer, Ferdinand von Richthofen. In reality, it was a vast, shifting web of caravan routes across deserts, mountains, and steppes, connected to sea lanes across the Indian Ocean. Very few merchants traveled the whole way. Goods passed from hand to hand, city to city, like a whisper passed along a line of children."

He stood up, a little stiffly, and pulled a large map from a shelf, unrolling it across the rug and weighing down the corners with books.

"Here, in the second century BCE, the Han emperor Wu sent an envoy named Zhang Qian westward, looking for allies against nomadic raiders. Zhang Qian was captured and held prisoner for about ten years. He married, had a child, escaped, kept going, and eventually returned home with astonishing reports of kingdoms in Central Asia and horses so strong and swift that people said they sweated blood. His journeys helped open China's trade with the west.

"And here," his finger moved across the map, "is Persia. Long before the Han, the Persian Achaemenid Empire, founded by Cyrus the Great in the sixth century BCE, stretched from the Aegean Sea to the Indus River. It built a Royal Road of about 2,700 kilometers, with relay stations where messengers could change horses. The Greek historian Herodotus admired those riders so much that a line based on his description was later carved over the main post office in New York City, praising couriers whom neither snow nor rain could stop.

"Persia also gave the world Nowruz, the 'new day,' a spring festival celebrated at the equinox for around three thousand years. Even today, hundreds of millions of people from Iran to Afghanistan, Central Asia, the Caucasus, and beyond celebrate it. Families set a table with seven items whose Persian names begin with the letter 's' — sprouts, garlic, apples, vinegar, and others — each a symbol of hope for the year ahead."

His finger moved again, to a city deep in Central Asia.

"Samarkand. For centuries, one of the great crossroads of the world. The Sogdian merchants who lived here were the great middlemen of the Silk Roads, speaking many languages, writing letters home on paper, and running trading networks that stretched across the continent. Along the routes there were caravanserais, roadside inns built around large courtyards, where camels could rest, merchants could sleep safely, and travelers from opposite ends of the world could share a meal."

"What did they trade besides silk?"

"Everything. Horses, jade, glass, gold, silver, spices, porcelain, paper, furs, wool, dates, pistachios. Ideas, too, which were the most valuable cargo of all. Buddhism traveled from India into Central Asia and then into China along these roads. So did music, mathematics, astronomy, and art styles. Paper-making traveled westward. Grapes, alfalfa, and new kinds of musical instruments traveled eastward. In the city of Dunhuang, at the edge of the desert, there are hundreds of cave temples, the Mogao Caves, painted over a thousand years with murals that mix Chinese, Indian, Persian, and Central Asian styles. You can see the whole world meeting on those walls.

"On the sea routes, ships rode the monsoon winds, which blow from the southwest in summer and from the northeast in winter. Sailors timed their voyages by them. Pepper and cardamom from the Malabar Coast, cinnamon from Sri Lanka, cloves and nutmeg from the Maluku Islands, fine cotton and muslin from Bengal and Gujarat, all moved across the Indian Ocean long before Europeans arrived there."

"But not everything that traveled was good," Elias continued, his voice softening. "In the 1340s, a terrible plague moved along the trade routes from Central Asia toward the Black Sea, and from there by ship to the Mediterranean. In 1347 it reached Europe. We now call it the Black Death. Within a few years, it killed perhaps a third or more of Europe's population, and it devastated the Middle East and North Africa as well. The same roads that carried silk and stories carried fleas and fear."

Noor shivered. "That's awful."

"It is. Connection always brings both gifts and dangers. That's as true now as it was then."

"Were there famous travelers?"

"Many. A Venetian named Marco Polo claimed to have traveled to China between 1271 and 1295 and to have served at the court of the Mongol ruler Kublai Khan. His book, dictated while he sat in a prison cell in Genoa, made Europeans dream of the East for centuries. Some historians have doubted parts of his story, but many of his details hold up surprisingly well.

"But my personal hero is Ibn Battuta. He was born in Tangier, in Morocco, in 1304. In 1325, at the age of twenty-one, he set out on pilgrimage to Mecca. He didn't come home for about twenty-four years. He traveled across North Africa, the Middle East, East Africa, Central Asia, India, the Maldives, Southeast Asia, and China. Later he crossed the Sahara to the Mali Empire, too. In all, he covered something like 117,000 kilometers, roughly 73,000 miles, far more than Marco Polo. He served as a judge in Delhi and in the Maldives, survived shipwrecks and bandits, and married several times along the way. When he finally settled down, the Sultan of Morocco asked him to dictate his adventures to a scholar. The result is called the Rihla, which simply means 'The Journey.'"

"I want to do that," Noor said firmly. "I want to travel for twenty-four years."

"Then you'd better start by learning some geography," said Elias, rolling up the map, "and finishing your homework."


CHAPTER FIVE: OLIVES, THEATERS, AND ROADS OF STONE

The next thing Noor picked was a small silver coin, so worn that its design was barely visible. When she tilted it toward the lamp, she could just make out the shape of an owl with enormous eyes.

"An Athenian tetradrachm," Elias said. "Or at least a very good replica. The owl was the symbol of Athena, goddess of wisdom and patron of the city. People across the Mediterranean simply called these coins 'owls.'"

"Like nicknames for money now."

"Exactly like that. Ancient Greece wasn't a single country. It was a collection of hundreds of independent city-states, each called a polis, scattered across mountains, islands, and coastlines, from what is now Spain to the Black Sea. They shared a language, gods, and festivals, but they fought one another about as often as they cooperated. Every four years, though, they came together at Olympia for games in honor of Zeus. According to tradition, the first Olympic Games were held in 776 BCE. There was a sacred truce so that athletes and spectators could travel safely. The modern Olympics, revived in Athens in 1896, borrowed that ancient idea."

"What was Athens famous for?"

"Many things, but above all an experiment. In 508 BCE, a statesman named Cleisthenes introduced reforms that gave power to the assembly of citizens. The Greeks called it demokratia: from demos, 'the people,' and kratos, 'power' or 'rule.' Every male citizen could speak and vote in the assembly. Many officials were chosen by lottery, using a clever stone device called a kleroterion, because the Athenians believed that elections favored the rich and famous, while a lottery gave ordinary people a fair chance."

"Was it fair, though?"

"Not by our standards. Women, enslaved people, and foreigners living in Athens couldn't vote, so only a minority of the adults in the city had a say. It's important to remember both things at once: it was a remarkable idea, and it was limited. Most of history is like that.

"The Athenians also loved theater. Thousands of people would gather in open-air theaters carved into hillsides to watch tragedies by Aeschylus, Sophocles, and Euripides, and comedies by Aristophanes, who loved making fun of politicians. The acoustics in some of these theaters were so good that, at Epidaurus, a coin dropped on the stage can supposedly be heard in the highest row of seats."

"And the philosophers?"

"Socrates, who wrote nothing down and spent his days asking people uncomfortable questions in the marketplace. He was put on trial and sentenced to death in 399 BCE, accused of corrupting the young and disrespecting the gods. His student Plato wrote dialogues starring Socrates and founded a school called the Academy. Plato's student Aristotle wrote about almost everything: logic, biology, ethics, politics, poetry, physics. And Aristotle's most famous student was a prince from Macedonia named Alexander."

"Alexander the Great?"

"That one. Between 334 and 323 BCE, he conquered an empire stretching from Greece to Egypt and all the way to the edge of India. He died in Babylon at the age of thirty-two. His empire broke apart almost at once, but Greek language and culture spread across the lands he had conquered. In Egypt, one of his generals founded a dynasty, the Ptolemies, that made the city of Alexandria into a center of learning. Its famous Library aimed to collect a copy of every book in the world. There, in the third century BCE, a scholar named Eratosthenes compared the angle of the sun's shadow in two different cities and calculated the circumference of the Earth. His answer was remarkably close to the true figure."

"Without satellites?"

"With a stick, a shadow, a well, and a very good brain."

Elias picked up a small piece of fired clay from another compartment: a fragment of roof tile stamped with Latin letters.

"Meanwhile, on the Italian peninsula, a small town was growing into something enormous. According to legend, Rome was founded in 753 BCE by twin brothers, Romulus and Remus, who had been raised by a she-wolf. In 509 BCE, the Romans threw out their last king and created a republic, governed by elected officials and a senate. Over the following centuries, Rome conquered all of Italy, then Carthage, then Greece, then much of Europe, North Africa, and the Middle East.

"After Julius Caesar was assassinated in 44 BCE, his adopted heir Octavian became the first emperor, taking the name Augustus in 27 BCE. At its greatest extent, in the second century CE, the empire may have held sixty million people or more. Rome built roads that ran straight across mountains and marshes, tens of thousands of miles of them, and aqueducts that carried fresh water for kilometers into cities. Romans perfected concrete; the dome of the Pantheon in Rome, completed around 125 CE, is still the largest unreinforced concrete dome in the world.

"Latin, the language of Rome, became the parent of Italian, Spanish, French, Portuguese, and Romanian. And English, though it isn't descended from Latin, borrowed thousands of words from it. Even this word I'm saying right now, 'history,' came through Latin from Greek."

"What happened to Pompeii?" Noor asked. "I read that a volcano buried it."

"On a summer or autumn day in 79 CE, Mount Vesuvius erupted. Pompeii and its neighbor Herculaneum were buried under ash, pumice, and superheated gas. It was a catastrophe for the people who lived there. But for historians, it froze a Roman town in a single moment: bread still in ovens, graffiti on the walls, election posters, advertisements, jokes, even rude drawings. One wall scribble, roughly translated, says the writer is amazed the wall hasn't collapsed under the weight of so much nonsense written on it."

Noor laughed out loud. "People were the same back then."

"That," said Elias, "is one of the most important lessons in all of history. People have always been people.

"In 476 CE, the last emperor in the West was removed from power, and historians often use that year to mark the end of the Western Roman Empire. But the Eastern half, which we call the Byzantine Empire, survived with its capital at Constantinople for almost another thousand years, until 1453."


CHAPTER SIX: THE HOUSE OF WISDOM

The object Noor chose next was heavier than the others: a round brass disk, intricately engraved, with a rotating pointer and a ring at the top so it could hang from a finger. Tiny numbers and curling letters covered every surface.

"What is it?"

"An astrolabe," said Elias, and his eyes shone. "This is one of my most treasured things. It's a kind of ancient computer, made entirely of brass. With it, you could tell the time by the sun or the stars, find the direction of Mecca for prayer, measure the height of a building, and work out when certain stars would rise. The basic idea goes back to the Greeks, but scholars in the Islamic world refined it into an instrument of extraordinary precision and beauty.

"In 762 CE, the Abbasid caliph al-Mansur founded a new capital on the banks of the Tigris River, not far from the ancient ruins of Babylon. He designed it as a perfect circle, with the palace and great mosque at its center. He called it Madinat al-Salam, the City of Peace. Everyone else called it Baghdad. Within a few decades it was one of the largest and wealthiest cities in the world.

"In Baghdad there was an institution called Bayt al-Hikma, the House of Wisdom. It flourished especially under the caliph al-Ma'mun in the early ninth century. Scholars there translated thousands of works from Greek, Persian, Syriac, and Sanskrit into Arabic: the medicine of Galen, the geometry of Euclid, the astronomy of Ptolemy, the philosophy of Aristotle, Indian mathematics. But they didn't just copy; they argued with, corrected, and built upon what they read.

"Paper made all of this easier. Paper-making had come westward from China into Central Asia, and the first paper mill in Baghdad was established in the 790s. Suddenly books were cheaper to make than ever before. Baghdad had a whole street of booksellers, and private libraries could hold thousands upon thousands of volumes.

"One of the most important scholars of the House of Wisdom was a Persian mathematician named Muhammad ibn Musa al-Khwarizmi. Around 820 CE, he wrote a book about solving equations. Its title included the Arabic word al-jabr, meaning 'restoring' or 'reunion of broken parts.' That's where we get the word algebra. And when his works were translated into Latin, his own name was written as Algoritmi. That's where we get the word algorithm."

Noor's eyes widened. "So every computer program..."

"...carries the name of a man from ninth-century Baghdad. Yes. His books also helped spread the Hindu-Arabic numerals, including zero, westward toward Europe.

"Then there was Ibn al-Haytham, born in Basra around 965. He wrote the Book of Optics, in which he argued, correctly, that we see because light enters our eyes, not because our eyes send out rays, as many ancient thinkers had believed. More importantly, he insisted on testing ideas through careful experiments rather than simply trusting famous authorities. Many historians consider him one of the pioneers of the scientific method. And Ibn Sina, known in Europe as Avicenna, completed his Canon of Medicine in 1025. It was used as a medical textbook in European universities until the seventeenth century.

"Let's not forget the poets, either: Rumi, whose poems about love and longing are still among the most widely read in the world, and Omar Khayyam, who was also a brilliant mathematician and astronomer and helped create a calendar more accurate than the one most of the world uses now.

"Far to the west, in the Iberian Peninsula, the region Muslims called al-Andalus, the city of Córdoba became one of the most brilliant cities in Europe during the tenth century. It had paved and lit streets, public baths, and a royal library said to hold hundreds of thousands of books. Its Great Mosque, begun in 785, is a forest of more than eight hundred columns topped by striped red and white arches. Muslims, Christians, and Jews lived side by side, not always peacefully, but often productively. The Jewish philosopher Maimonides was born in Córdoba in 1138, and the Muslim philosopher Ibn Rushd, known in Europe as Averroes, was born there in 1126. His commentaries on Aristotle shaped European philosophy for centuries.

"After Christian kingdoms captured Toledo in 1085, scholars gathered there to translate Arabic books into Latin. Through those translations, Europe rediscovered much of ancient Greek learning, along with the new mathematics, medicine, and astronomy of the Islamic world. It was one of the greatest relays of knowledge in history.

"And in Granada, the last Muslim kingdom on the peninsula, the Nasrid rulers built the Alhambra, a palace complex of courtyards, fountains, and walls covered in carved plaster so delicate it looks like lace. Water flows through it everywhere, cooling the air. Granada fell to the Catholic Monarchs, Ferdinand and Isabella, in January 1492."

"What happened to Baghdad?"

"In 1258, the Mongol army sacked the city. It's said the Tigris ran dark with ink from all the books thrown into it. The details may be exaggerated, but the destruction was real and terrible. Yet ideas, once shared, are hard to kill. The knowledge of the House of Wisdom had already spread across three continents."

Noor ran her thumb over the engraved letters on the astrolabe. "It's strange," she said. "At school they make it sound like knowledge went straight from Greece to Rome to Europe. But it went around."

"It always goes around," said Elias. "Knowledge is never a straight line. It's a conversation."


CHAPTER SEVEN: GOLD, SALT, AND MANUSCRIPTS

From the chest Noor lifted a small, rough, grayish-white block wrapped in cloth, and beside it a single page of old paper covered in elegant brown Arabic script.

"Salt," said Elias, "and a copy of a page from a manuscript from Timbuktu. In West Africa, several hundred years ago, these two things — salt and books — were worth more than you can imagine."

"Salt? Like on chips?"

"Like on chips. In the hot climate of West Africa, people and animals needed salt to survive, but the lands south of the Sahara had little of it. Far to the north, in the desert, there were great deposits of rock salt at places like Taghaza. Meanwhile, the lands of the south had gold, a great deal of it. So camel caravans, sometimes with thousands of animals, crossed the Sahara carrying salt south and gold north. In some markets, it was said, salt was traded weight for weight with gold.

"The first great empire to grow rich on this trade was Ghana, which its people called Wagadu. It flourished from around the fourth century until the thirteenth. It wasn't in the same place as the modern country of Ghana, which took its name in honor of the old empire when it gained independence in 1957. Arab geographers wrote that the king of Ghana held court surrounded by dogs wearing gold collars.

"After Ghana declined, a new empire rose: Mali. Its founder was Sundiata Keita, who defeated a powerful sorcerer-king named Sumanguru around 1235. Sundiata's life is told in an epic that has been passed down for nearly eight hundred years by griots."

"What's a griot?"

"A griot, or jeli in the Mandinka language, is a storyteller, historian, musician, and keeper of memory, all at once. Griots memorize the genealogies of families and the deeds of kings and recite them to the music of the kora, a beautiful instrument with twenty-one strings and a body made from a large calabash gourd. In a world where many people didn't read, griots were living libraries. Many still perform today.

"The most famous ruler of Mali was Mansa Musa, who came to power in the early 1300s. In 1324 he made the pilgrimage to Mecca. Accounts say he traveled with thousands of attendants and dozens of camels loaded with gold. When he passed through Cairo, he spent and gave away so much gold that its value in Egypt fell, and Egyptian writers were still talking about it twelve years later. Some modern writers have called him the richest person who ever lived, though it's impossible to measure that properly.

"In 1375, a mapmaker on the island of Majorca drew Mansa Musa on a famous map, the Catalan Atlas. He's shown seated on a throne, wearing a golden crown and holding up a golden nugget. Europe had heard of Mali's wealth.

"But Mansa Musa cared about more than gold. He returned from Mecca with scholars and an architect, and he commissioned the Djinguereber Mosque in Timbuktu, built of mud brick, which still stands. Timbuktu became one of the greatest centers of learning in the Islamic world. Scholars there studied law, theology, astronomy, mathematics, medicine, grammar, and poetry. Families kept private libraries of handwritten manuscripts, passed down for generations.

"In 2012, when armed militants occupied Timbuktu and threatened its heritage, librarians and families secretly smuggled hundreds of thousands of manuscripts out of the city, packed in metal trunks and rice sacks, on donkey carts, in cars, and on river boats, to safety in the capital, Bamako. Some manuscripts were burned, but the vast majority were saved by ordinary people who loved books."

Noor held the manuscript page very carefully now.

"And Africa is enormous," Elias continued. "Far bigger than most maps make it look. You could fit the United States, China, India, and most of Europe inside it, with room to spare. So there are many, many other stories. In what is now Nigeria, the Nok culture made remarkable terracotta sculptures of human heads more than two thousand years ago. In the kingdom of Benin, guilds of artists cast brass plaques and sculptures of extraordinary skill. In 1897, British soldiers looted thousands of these Benin Bronzes, and today museums around the world are debating, and in some cases carrying out, their return.

"In the highlands of Ethiopia, the kingdom of Aksum minted its own coins, traded across the Red Sea, raised giant carved stone stelae, and adopted Christianity in the fourth century, making it one of the earliest Christian states in the world. Ethiopia is also, according to a popular legend, the place where a goat herder named Kaldi noticed his goats dancing after eating the red berries of a certain bush. Those were coffee berries. The legend probably isn't true, but coffee really does come from Ethiopia, and it was in Yemen, just across the Red Sea, that people began brewing it as a drink.

"On the east coast, cities like Kilwa, Mombasa, and Zanzibar grew rich from Indian Ocean trade, linking Africa with Arabia, Persia, India, and China. Their Swahili culture blended African, Arab, and Persian influences, and the Swahili language, a Bantu language with many Arabic loanwords, is now spoken by tens of millions of people. And in southern Africa, between the eleventh and fifteenth centuries, the people of Great Zimbabwe built massive stone walls, some more than ten meters high, without any mortar at all. The stones simply fit together. The modern nation of Zimbabwe takes its name from that city: in the Shona language, dzimba dza mabwe means 'houses of stone.'"

"There's a saying," Elias said quietly, "that the writer Amadou Hampâté Bâ, from Mali, made famous at UNESCO in 1960: in Africa, when an old person dies, a library burns. It's why griots matter. And it's why I'm telling you these stories."

Noor looked up at him and didn't say anything for a long moment.


CHAPTER EIGHT: CORN, CALENDARS, AND KNOTTED STRINGS

Next Noor found a bundle of colored cotton cords, all hanging from one main cord, each tied with knots of different shapes at different heights.

"That's a quipu," said Elias. "Or khipu. It's a replica, made for me by a weaver in Peru. The Inca Empire used these knotted strings to keep records: census data, taxes, stores of food, perhaps even stories and histories. The type of knot, its position, and the color of the cord all carried meaning. Specialists called quipucamayocs could read them. Scholars have worked out how many quipus recorded numbers, using a base-ten system, but we still don't fully understand all of them."

"Writing with knots," Noor murmured. "That's so clever."

"The Americas are full of clever. For thousands of years they developed entirely independently from Europe, Africa, and Asia, and people there invented agriculture, cities, writing, mathematics, and monumental architecture on their own. Around nine thousand years ago, farmers in what is now southern Mexico began domesticating a wild grass called teosinte, which had tiny ears with only a few hard kernels. Over thousands of years of careful breeding, they turned it into maize, or corn. It's one of the greatest feats of plant breeding in history.

"In fact, a huge share of the food the world eats today came from the Americas: maize, potatoes, tomatoes, chili peppers, peanuts, beans, squash, pumpkins, avocados, vanilla, and chocolate. Imagine Italian food without tomatoes, Indian food without chilies, or Irish history without potatoes. None of those things existed outside the Americas before 1492.

"The Olmec civilization, along the Gulf Coast of Mexico, flourished from around 1200 BCE. They carved enormous stone heads, some nearly three meters tall and weighing many tons, probably portraits of their rulers. Seventeen have been found so far.

"Later came the Maya, in southern Mexico and Central America. Their Classic period lasted from about 250 to 900 CE. They built cities like Tikal, Palenque, and Copán, with towering stepped pyramids and ball courts where a ritual game was played with a heavy rubber ball. Yes, rubber. People in Mesoamerica were making rubber balls more than three thousand years ago, by mixing latex from rubber trees with the juice of morning glory vines.

"The Maya developed the most complete writing system in the ancient Americas, with hundreds of signs. For a long time, scholars couldn't read it, but in the second half of the twentieth century a breakthrough came, partly thanks to a Russian linguist named Yuri Knorozov, who argued that many Maya glyphs represented syllables. Today most Maya inscriptions can be read. The Maya were also brilliant astronomers and mathematicians. They used a concept of zero, and their calendar, the Long Count, could track enormous spans of time. Some people believed the Maya predicted the world would end in December 2012, when one great cycle of the calendar ended. They didn't. It was more like a car's odometer rolling over. And millions of Maya people still live in the region today, speaking around thirty Mayan languages.

"Near today's Mexico City was Teotihuacan, a vast city with a Pyramid of the Sun and a Pyramid of the Moon connected by a long avenue. At its height, around 500 CE, it may have had a hundred thousand residents or more. Centuries after it was abandoned, the Aztecs found its ruins and gave it the name we use, which means something like 'the place where gods were made.'

"The Aztecs, or Mexica, founded their own capital, Tenochtitlan, around 1325, on an island in Lake Texcoco. Their legend said their god had told them to settle where they saw an eagle perched on a cactus. That image is on the flag of Mexico today. They expanded the island with chinampas, artificial garden plots built up from mud in the shallow lake, which were wonderfully productive. When Spanish soldiers arrived in 1519, they were astonished. Tenochtitlan may have had two hundred thousand people, larger than any city in Spain at the time, with causeways, aqueducts, canals, and a vast market.

"High in the Andes mountains of South America, the Inca built the largest empire in the Americas in less than a century. They called it Tawantinsuyu, 'the four parts together.' It stretched about four thousand kilometers along the western side of South America. The Inca built a road network of some thirty to forty thousand kilometers, the Qhapaq Ñan, with rope bridges across deep gorges and relay runners called chasquis who could carry messages hundreds of kilometers in a day. Their stonework was so precise that, in some walls in Cusco, you can't slide a knife blade between the stones. And around 1450, they built Machu Picchu, high on a mountain ridge. The Spanish never found it."

"And north of Mexico?"

"Many peoples. Near what is now St. Louis, from around 1050 to 1350 CE, there was a city called Cahokia, with an earthen pyramid, Monks Mound, that is the largest prehistoric earthwork in the Americas. In the northeast, five nations, the Mohawk, Oneida, Onondaga, Cayuga, and Seneca, formed the Haudenosaunee Confederacy, bound together by the Great Law of Peace. A sixth nation, the Tuscarora, joined in 1722. Women, especially clan mothers, had the power to choose and remove chiefs. Some historians argue that the confederacy influenced the thinking of the founders of the United States."

Elias paused, and his face grew serious.

"I have to tell you the hard part too. When Europeans arrived after 1492, they brought diseases like smallpox and measles, against which the peoples of the Americas had no immunity. Combined with war, enslavement, and the destruction of communities, these epidemics killed a very large proportion of the Indigenous population, in some places as many as nine out of ten people. It was one of the greatest demographic catastrophes in human history."

"That's horrible."

"It is. But it's not the end of the story. Indigenous peoples across the Americas survived. They're still here, tens of millions of them, keeping languages, ceremonies, and knowledge alive. Quechua, the language of the Inca, is spoken by millions of people today. In Mexico, families celebrate Día de los Muertos, the Day of the Dead, at the beginning of November, building altars called ofrendas decorated with marigolds, candles, photographs, and the favorite foods of loved ones who've died. It blends Indigenous and Catholic traditions. It isn't a sad holiday. It's a joyful reunion with memory."

"I like that," said Noor softly. "Remembering people with their favorite food."

"Me too. When my time comes, I'd like plenty of strong coffee and a slice of lemon cake."

"Grandfather!"

"I'm only planning ahead," he said, eyes twinkling.


CHAPTER NINE: STARS, SWELLS, AND STONE TEMPLES

The seashell turned out to be tied to a small frame of thin, crisscrossing sticks, with other shells fastened at the places where the sticks met.

"A stick chart," Elias said, "from the Marshall Islands, in the Pacific Ocean. The curved sticks show how ocean swells bend around islands, and the shells mark the islands themselves. Navigators studied charts like this on land, memorized them, and then left them at home. Out at sea, they read the waves with their bodies, sometimes by lying down in the canoe and feeling the swells through the hull."

"Without a compass?"

"Without a compass, a map, or a single metal instrument. More than three thousand years ago, people of the Lapita culture began sailing east from the islands near New Guinea into the open Pacific. Their descendants, the Polynesians, became the greatest ocean voyagers the world had ever seen. In double-hulled canoes, they found and settled a triangle of islands spread across an area larger than all of Asia: Hawaiʻi in the north, Rapa Nui, or Easter Island, in the east, and Aotearoa, which most people now call New Zealand, in the southwest, settled around 1250 to 1300 CE. They steered by the rising and setting points of stars, by the shape of the swells, by the flight of seabirds returning to land at dusk, and by the color of clouds above distant lagoons.

"For a long time, some European scholars refused to believe this was possible. They claimed the islands must have been settled by accident, by canoes drifting with the wind. So in 1976, a group in Hawaiʻi built a traditional voyaging canoe called Hōkūleʻa. A master navigator from Micronesia named Mau Piailug guided it all the way from Hawaiʻi to Tahiti, more than four thousand kilometers, without any instruments at all. Between 2014 and 2017, Hōkūleʻa sailed around the entire world. The old knowledge had been kept alive just long enough to be taught again."

He turned the stick chart gently in the lamplight.

"On Rapa Nui, the islanders carved nearly a thousand giant stone figures called moai, to honor their ancestors. In Southeast Asia, kingdoms grew rich on the sea trade between India and China. On the island of Java, in the ninth century, builders created Borobudur, a Buddhist monument shaped like a mountain, with more than two and a half thousand carved relief panels that pilgrims follow as they climb toward the top. In Cambodia, in the twelfth century, the Khmer king Suryavarman II built Angkor Wat, still the largest religious monument in the world. It was first dedicated to the Hindu god Vishnu and later became a Buddhist temple, and today it appears on Cambodia's flag.

"And to the north, in Japan, around the year 1008, a lady of the imperial court named Murasaki Shikibu wrote The Tale of Genji, a long story about love, politics, and the passing of time. Many people call it one of the world's first novels. Japan also developed the tea ceremony, perfected by a tea master named Sen no Rikyū in the sixteenth century, in which every movement, from folding the cloth to turning the bowl, is done with complete attention. It's connected to an idea called wabi-sabi: finding beauty in things that are simple, imperfect, and impermanent."

"Like an old shell," said Noor.

"Exactly like an old shell."


CHAPTER TEN: PRESSES, ENGINES, AND A TICKET HOME

Near the bottom of the chest lay a single sheet of thick paper printed in dense black letters, and beneath it a folded railway ticket, yellow with age.

"A facsimile of a page from the Gutenberg Bible," said Elias. "Around 1440, in the German city of Mainz, a goldsmith named Johannes Gutenberg developed a printing press with movable metal type and oil-based ink. Bi Sheng in China and printers in Korea had used movable type centuries earlier, but Gutenberg's system was fast and suited perfectly to an alphabet of only a few dozen letters. His Bible was finished around 1455. By the year 1500, presses in hundreds of European towns had printed millions of books. Ideas could now spread faster than any king or church could control them.

"That helped fuel the Renaissance, a 'rebirth' of interest in ancient learning that began in Italian cities like Florence. Leonardo da Vinci filled thousands of notebook pages with sketches of flying machines, human anatomy, and swirling water. Michelangelo spent four years, from 1508 to 1512, painting the ceiling of the Sistine Chapel. In 1517, a monk named Martin Luther challenged the Church, and printed pamphlets carried his arguments across Europe within weeks. The Reformation split Western Christianity in two.

"Farther east, in 1453, the Ottoman sultan Mehmed II captured Constantinople, ending the Byzantine Empire, and the city, later known as Istanbul, became the heart of an empire that stretched across three continents. In the 1550s, Istanbul's first coffeehouses opened, serving the drink that had come from Yemen. People gathered there to play games, listen to storytellers, read poetry aloud, and argue about politics, so much so that some called them schools of the wise. The idea spread to Venice, Vienna, Paris, and London. In the 1680s, merchants and ship owners began meeting at Edward Lloyd's coffeehouse in London to share news about ships and cargo, and that meeting place grew into Lloyd's, one of the most famous insurance markets in the world. A cup of coffee, it turns out, can change history.

"In 1492, Christopher Columbus crossed the Atlantic, and the two halves of the world were joined in what historians call the Columbian Exchange: horses, wheat, and sugar went west; maize, potatoes, and tomatoes went east. But the exchange also carried enslaved people. Over about four centuries, more than twelve million Africans were forced across the Atlantic in chains. European empires grew rich, and many peoples across Africa, Asia, and the Americas lost their land, their freedom, and their right to govern themselves."

Noor unfolded the railway ticket. "And this?"

"My grandfather's ticket, from a journey he took as a young man. Railways are part of the Industrial Revolution, which began in Britain in the eighteenth century, when steam engines and machines began doing work that had always been done by hand. In 1830, the Liverpool and Manchester Railway opened, the first intercity line in the world to run every service with steam locomotives. Within a few decades, railways crossed whole continents. Distances that had taken weeks suddenly took days.

"It was also an age of people demanding freedom. In 1804, Haiti became the first nation founded after a successful revolt of enslaved people. Slavery was abolished across the British Empire in the 1830s, in the United States in 1865, and in Brazil, the last country in the Americas to end it, in 1888. In the twentieth century, after two terrible world wars, colony after colony won independence: India and Pakistan in 1947, Ghana in 1957, and seventeen African nations in 1960 alone, which is sometimes called the Year of Africa. In 1945, the United Nations was founded, and in 1948 it adopted the Universal Declaration of Human Rights, which has since been translated into more than five hundred languages.

"And then, in 1989, a British scientist named Tim Berners-Lee, working at CERN in Switzerland, proposed what became the World Wide Web. Today, a girl in a rainy harbor town can read about Timbuktu, Teotihuacan, and Tahiti in the time it takes to boil a kettle."

"Then why do we need the chest?" Noor asked.

Elias smiled. "Because of all the world's roughly seven thousand languages, around forty percent are endangered. Some have only a handful of speakers left. Every one of them holds songs, jokes, recipes, and ways of seeing that exist nowhere else. A search engine can't remember those for us. People have to."


EPILOGUE: THE EMPTY COMPARTMENT

Outside, the rain had finally stopped. A thin line of gold showed beneath the clouds over the harbor, and gulls had begun to call again.

Noor looked down into the chest. Most of the compartments were full once more, each object back on its faded velvet. But in the very last row there was one compartment she hadn't noticed before. It was empty. Its paper label, in her grandfather's careful, slanting hand, said simply: For Noor.

"What goes in there?" she asked.

"Whatever you choose," said Elias. "Something from a place you've been, or a person you've met, or a story you've heard and want to keep. History isn't only kings and pyramids, Noor. It's farmers keeping track of barley, a man complaining about bad copper, a girl writing a novel at a royal court, a navigator reading the waves, a family hiding manuscripts in rice sacks. It's ordinary people deciding what's worth remembering."

Noor thought about it for a long time. Then she ran to her room and came back with a small, smooth gray stone she'd found on the beach the summer before, the one with a perfect white stripe around its middle.

"It's from here," she said, setting it in the compartment. "From Port Selene. Somebody should remember this place too."

Elias took off his spectacles and wiped his eyes, though he claimed it was only dust.

"Then tell me its story," he said.

And she did.

THE END
"""
```

### Text preprocessing

```python
def sanitized_tokens(text):
    text = text.lower()
    text = re.sub(r"[\"']", "", text)

    # remove punctuations
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # tokenize
    tokens = word_tokenize(text)

    return tokens
```

```python

vocab = {'<unk>': 0}

for token in Counter(tokens).keys():
    if token not in vocab:
        vocab[token] = len(vocab)

```

#### Text to indices

```python
def text_to_indices(sentence, vocab):
    numerical_sentences = []
    for token in sentence:
        if token in vocab:
            numerical_sentences.append(vocab[token])
        else:
            numerical_sentences.append(vocab['<unk>'])
    return numerical_sentences
```

### LSTM Model

```python

class LSTMModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(num_embeddings=vocab_size, embedding_dim=64)
        self.lstm = nn.LSTM(input_size=64, hidden_size=128, batch_first=True)
        self.fc = nn.Linear(in_features=128, out_features=vocab_size)

    def forward(self, X):
        embedded = self.embedding(X)
        intermediate_hidden_state, (final_hidden_states, final_cell_states) = self.lstm(embedded)
        output = self.fc(final_hidden_states.squeeze(0))

        return output

```

```python
model = LSTMModel(vocab_size=len(vocab))
```

```python
model.to(device='cpu')
```

#### Training Loop

```python
epochs = 50
learning_rate = 0.001
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
```

```python

for epoch in range(epochs):
    model.train()

    total_loss = 0.0

    for batch_x, batch_y in train_loader:
        optimizer.zero_grad()

        output = model(batch_x)
        loss = criterion(output, batch_y)

        loss.backward()
        optimizer.step()

        total_loss = total_loss + loss.item()

    print(f'Epoch: {epoch+1} - Total loss: {total_loss}')
```

#### Predictions

```python
def prediction(model, vocab, text):
    tokenized_text = sanitized_tokens(text)
    numerical_text = text_to_indices(tokenized_text, vocab)
    padded_numerical_text = [0] * (max(len_list) - len(numerical_text)) + numerical_text

    padded_numerical_text = torch.tensor(padded_numerical_text, dtype=torch.long).unsqueeze(0)

    output = model(padded_numerical_text)
    _, predicted_idx = torch.max(output, dim=1)
    predicted_text = list(vocab.keys())[predicted_idx]

    complete_predicted_seq = text + " " + predicted_text

    return complete_predicted_seq
    
```

__Accuracy__

```python
def calculate_accuracy(model, dataloader, device):
    model.eval()

    correct = 0
    total = 0

    with torch.no_grad():
        for batch_x, batch_y in dataloader:
            output = model(batch_x)
            _, predicted = torch.max(output, dim=1)

            correct += (predicted == batch_y).sum().item()
            total += batch_y.size(0)

    accuracy = correct / total

    return accuracy
```

```python

train_accuracy = calculate_accuracy(model, train_loader, device)
test_accuracy = calculate_accuracy(model, test_loader, device)

print(f'Model train accuracy: {train_accuracy:.2f}')
print(f'Model test accuracy: {test_accuracy:.2f}')
```
