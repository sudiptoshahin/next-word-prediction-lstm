### Text preprocessing

```python
def sanitized_tokens(text):
    text = text.lower()
    text = re.sub(r"[\"']", "", text)

    # remove punctuations
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # tokenize
    tokens = word_tokenize(text)

    return tokens
```

```python

vocab = {'unk': 0}

for token in Counter(tokens).keys():
    if token not in vocab:
        vocab[token] = len(vocab)

```

#### Text to indices

```python
def text_to_indices(sentence, vocab):
    numerical_sentences = []
    for token in sentence:
        if token in vocab:
            numerical_sentences.append(vocab[token])
        else:
            numerical_sentences.append(vocab['<unk>'])
    return numerical_sentences
```

```python
input_numerical_sentences = []
for sentence in input_sentences:
    tokenized_sentence = sanitized_tokens(sentence.lower())
    input_numerical_sentences.append(text_to_indices(tokenized_sentence, vocab))
```

for next word prediction

```python
training_sequences = []

for sentence in input_numerical_sentences:
    for i in range(1, len(sentence)):
        training_sequences.append(sentence[:i+1])
```

#### Padding

```python
len_list = []
for sequence in training_sequences:
    len_list.append(len(sequence))

max(len_list)
```

```python
# zero padding

padded_training_sequences = []
for sequence in training_sequences:
    padded_sequence = [0] * (max(len_list) - len(sequence)) + sequence
    padded_training_sequences.append(padded_sequence)
```

#### Convert to tensor

```python
class NextWordPredictionDataset(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```

### LSTM Model

```python

class LSTMModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(num_embeddings=vocab_size, embedding_dim=100)
        self.lstm = nn.LSTM(input_size=100, hidden_size=200, batch_first=True)
        self.fc = nn.Linear(in_features=200, out_features=vocab_size)

    def forward(self, X):
        embedded = self.embedding(X)
        intermediate_hidden_state, (final_hidden_states, final_cell_states) = self.lstm(embedded)
        output = self.fc(final_hidden_states.squeeze(0))

        return output

```
